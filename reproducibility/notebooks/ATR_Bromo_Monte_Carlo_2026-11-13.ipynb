# Bromo assumption-conditioned Monte Carlo

10,000 draws, seed 20261113; Python/NumPy/Matplotlib. These are **assumed independent distributions**, not empirical confidence intervals or a coupled seven-subsystem Monte Carlo. Thresholds: first annual-average admissions >= 2,752/day, and first year-end cultural index <60. Non-crossing trials remain censored at 2045 and are not assigned a fictitious 2046 date. See ../docs/BROMO_MONTE_CARLO_2026-11-13.md.

```python
from pathlib import Path
import sys
cwd = Path.cwd().resolve()
root = next((p for p in (cwd, *cwd.parents) if (p/'reproducibility/scripts/bromo_monte_carlo_assumptions.py').is_file()), None)
if root is None:
    raise FileNotFoundError('Run notebook from cloned repository')
sys.path.insert(0, str(root/'reproducibility/scripts'))
from bromo_monte_carlo_assumptions import simulate, describe, plot_envelopes
N, SEED = 10000, 20261113
output = root/'reproducibility/outputs/monte-carlo-2026-11-13'
output.mkdir(parents=True, exist_ok=True)
draws, trajectories = simulate(N, SEED, distribution='triangular')
summary = describe(draws)
for label, result in summary.items():
    print(label, result)
plot_envelopes(trajectories, output/'assumption_conditioned_mc_envelopes.png')
```

```python
from IPython.display import display, Image
import pandas as pd
base_draws, base_traj = simulate(1, 0, stochastic=False)
base = describe(base_draws)
assert base['bau_visitor']['earliest_crossing_year'] == 2040
assert base['bau_culture']['earliest_crossing_year'] == 2032
assert base['intervention_visitor']['not_crossed_by_2045'] == 1
assert base['intervention_culture']['not_crossed_by_2045'] == 1
display(pd.DataFrame(summary).T)
display(Image(filename=str(output/'assumption_conditioned_mc_envelopes.png')))
```
